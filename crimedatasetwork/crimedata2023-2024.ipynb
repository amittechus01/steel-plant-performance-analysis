 # ***<P style="font-family:newtimeroman;font-size:150%;text-align:center;color:green;"> Missing Values Analysis and Imputation </p>***

```python
df['Weapon Desc'].fillna("Unknown",inplace=True)
```

```python
df['Vict Age'] = df['Vict Age'].replace(0, np.nan)

df['Vict Age'] = df['Vict Age'].fillna(df['Vict Age'].median())



df['Vict Sex'].fillna(df['Vict Sex'].mode()[0], inplace=True)

df['Location Type'].fillna(df['Location Type'].mode()[0], inplace=True)


df['Case Status'].fillna(df['Case Status'].mode()[0], inplace=True)


df.loc[df['Vict Age'] < 0, 'Vict Age'] = np.nan  
df['Vict Age'] = df['Vict Age'].fillna(df['Vict Age'].median())
```

```python
df.isnull().sum()
```

 # ***<P style="font-family:newtimeroman;font-size:150%;text-align:center;color:green;"> let's Handle the space in all columns </p>***

```python
df=df.apply(lambda x: x.str.strip() if x.dtype=='object' else x)
```

```python
df.isnull().sum()
```

 # ***<P style="font-family:newtimeroman;font-size:150%;text-align:center;color:green;">All missing values handled so we change the the data types of columns according to your data </p>***

```python

df['Date Reported'] = pd.to_datetime(
    df['Date Reported']
    
)
```

```python
df['Date Occ'] = pd.to_datetime(
    df['Date Occ']
    
)
```

```python
df[['Date Reported','Date Occ']].head(5)
```

```python
df['Time Occ'] = (
    df['Time Occ']
    .astype(str)                
    .str.replace(':', '')       
    .str.strip()                 
    .str.zfill(4)               
)

df['Time Occ'] = df['Time Occ'].str[:2] + ':' + df['Time Occ'].str[2:]
```

```python
df['Time Occ'].head(5)
```

```python
df['Vict Age'] = df['Vict Age'].astype('int64')
```

```python
df.info()
```

# ***<p style="font-family:newtimeroman;font-size:200%;text-align:center;color:green;">Data Visualization</p>***

# ***<p style="font-family:newtimeroman;font-size:120%;text-align:center;color:green;">BAR PLOT</p>***

```python
Gender_count = df['Vict Sex'].value_counts()

plt.bar(Gender_count.index, Gender_count.values)

plt.title("Victim Gender Category")
plt.xlabel("Victim Gender")
plt.ylabel("Count")

plt.show()
```

```python
age_count = df['Vict Age'].value_counts()

plt.bar(age_count.index, age_count.values)

plt.title("Victim Age Distribution")
plt.xlabel("Victim Age")
plt.ylabel("Number of Victims")

plt.show()
```

```python
top_locations = df['Location Type'].value_counts().head(20)

plt.figure(figsize=(15,10))

sns.barplot(
    x=top_locations.values,
    y=top_locations.index
)

plt.title("Top Crime Location Types")
plt.xlabel("Number of Crimes")
plt.ylabel("Location Type")

plt.show()
```

# ***<p style="font-family:newtimeroman;font-size:150%;text-align:center;color:green;">PIE CHART</p>***

```python
area_count = df['Area Name'].value_counts().head(10)

plt.figure(figsize=(10,8))

plt.pie(
    area_count.values,
    labels=area_count.index,
    autopct='%1.1f%%'
)

plt.title("Top 10 Crime Area")
plt.show()
```

```python

hour = df['Time Occ'].astype(str).str[:2]
hour_count = hour.value_counts().sort_index()

plt.figure(figsize=(10,5))
plt.bar(hour_count.index, hour_count.values)

plt.title("Crime Occurrence by Hour (When Crimes Happen Most)")
plt.xlabel(" (00 = 12 AM, 23 = 11 PM)")
plt.ylabel("Number of Crimes  Each Hour")
plt.show()
```

```python
weapon = df['Weapon Desc'].value_counts().head(6)

plt.figure(figsize=(8,10))

plt.pie(
    weapon.values,
    labels=weapon.index,
    autopct='%1.1f%%'
)

plt.title("Weapon Used in Crime")
plt.show()
```

```python
area_count = df['Case Status'].value_counts()

area_count.plot(kind='bar')

plt.title("Case Status Distribution")
plt.xlabel("Case Status")
plt.ylabel("Total Count")
plt.xticks(rotation=45)
plt.show()
```

```python
CrimeType = df['Crime Desc'].value_counts().head(10)

plt.figure(figsize=(10,10))

plt.pie(
    CrimeType.values,
    labels=CrimeType.index,
    autopct='%1.1f%%'
)

plt.title("Top 10 Crimes")
plt.show()
```

# ***<p style="font-family:newtimeroman;font-size:150%;text-align:center;color:green;">DISTICT WISE CASE STATUS </p>***

```python
pivot = pd.crosstab(df['Rpt Dist No'], df['Case Status'])

pivot = pivot.head(20)

plt.figure(figsize=(12,8))

sns.heatmap(
    pivot,
    annot=True,
    fmt='d',
    cmap='coolwarm'
)

plt.title("District Wise top 20 Case Status")
plt.xlabel("Case Status")
plt.ylabel("District No ")

plt.show()
```

# ***<p style="font-family:newtimeroman;font-size:150%;text-align:center;color:green;"> CASE REPORTING STATUS </p>***

```python
delay = (df['Date Reported'] - df['Date Occ']).dt.days

delay_cat = pd.cut(
    delay,
    bins=[-1, 0, 2, 7, 1000],
    labels=['Same Day', '1-2 Days', '3-7 Days', '1 Week+']
)

values = delay_cat.value_counts()

plt.figure(figsize=(7,7))

plt.pie(values, labels=values.index, autopct='%1.1f%%', startangle=90)

plt.title("Case Reporting Status")
plt.axis('equal')

plt.show()
```

# ***<p style="font-family:newtimeroman;font-size:150%;text-align:center;color:green;"> DAY WISE CRIME RATE </p>***

```python

day = df['Date Occ'].dt.day_name()

values = day.value_counts()


plt.figure(figsize=(8,8))

plt.pie(values, labels=values.index, autopct='%1.1f%%', startangle=90)

plt.title("Day-wise Crime Distribution")
plt.axis('equal')

plt.show()
```
